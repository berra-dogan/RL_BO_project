```python
import pandas as pd
```

```python
df = pd.read_csv("lofo_comparison.csv")
print(df.columns)
df
```

```python
display_cols = [
    "reward",
    "final_regret",
    "mean_total_scaled_move_cost",
    "mean_decision_time",
]

for dim in sorted(df["dimension"].unique()):
    for horizon in sorted(df["horizon"].unique()):
        for f in sorted(df["held_out_function"].unique()):
            sub = df[
                (df["dimension"] == dim)
                & (df["horizon"] == horizon)
                & (df["held_out_function"] == f)
            ]
            if sub.empty:
                continue
            print(f"dimension={dim}, horizon={horizon}, held_out_function={f}")
            display(sub[display_cols].sort_values("final_regret"))
```

```python
# --- Compare horizon = 3 vs horizon = 5 ---
# Pivots the per-(dimension, held_out_function, reward) metrics so h=3 and h=5 sit
# side by side, then keeps only the raw delta (h5 - h3) and percent change.
# Negative regret delta = longer horizon helped.
compare_metrics = [
    "final_regret",
    "best_regret",
    "mean_total_scaled_move_cost",
    "mean_decision_time",
]
key_cols = ["dimension", "held_out_function", "reward"]

_h = df[df["horizon"].isin([3, 5])]
horizon_compare = _h.pivot_table(index=key_cols, columns="horizon", values=compare_metrics)
horizon_compare.columns = [f"{metric}_h{horizon}" for metric, horizon in horizon_compare.columns]
horizon_compare = horizon_compare.reset_index()

diff_cols = []
for metric in compare_metrics:
    c3, c5 = f"{metric}_h3", f"{metric}_h5"
    if c3 in horizon_compare and c5 in horizon_compare:
        horizon_compare[f"{metric}_delta"] = horizon_compare[c5] - horizon_compare[c3]
        horizon_compare[f"{metric}_pct_change"] = (
            100 * (horizon_compare[c5] - horizon_compare[c3]) / horizon_compare[c3].abs()
        )
        diff_cols += [f"{metric}_delta", f"{metric}_pct_change"]

horizon_compare = horizon_compare[key_cols + diff_cols]

# Per (dimension, held_out_function) table, biggest final-regret improvement first
for dim in sorted(horizon_compare["dimension"].unique()):
    for f in sorted(horizon_compare["held_out_function"].unique()):
        sub = horizon_compare[
            (horizon_compare["dimension"] == dim)
            & (horizon_compare["held_out_function"] == f)
        ]
        if sub.empty:
            continue
        print(f"dimension={dim}, held_out_function={f}")
        display(sub.drop(columns=["dimension", "held_out_function"]).sort_values("final_regret_delta"))

# Aggregate: mean delta / pct change per reward (across dimensions / functions)
print("Mean (h5 - h3) delta and pct change per reward:")
display(
    horizon_compare.groupby("reward")[diff_cols]
    .mean()
    .sort_values("final_regret_delta")
)
```
