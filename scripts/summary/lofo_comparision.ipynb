```python
import pandas as pd
```

```python
df = pd.read_csv("lofo_comparison.csv")
print(df.columns)
df
```

```python
display_cols = [
    "reward",
    "final_regret",
    "mean_total_scaled_move_cost",
]

for dim in sorted(df["dimension"].unique()):
    for horizon in sorted(df["horizon"].unique()):
        for f in sorted(df["held_out_function"].unique()):
            sub = df[
                (df["dimension"] == dim)
                & (df["horizon"] == horizon)
                & (df["held_out_function"] == f)
            ]
            if sub.empty:
                continue
            print(f"dimension={dim}, horizon={horizon}, held_out_function={f}")
            display(sub[display_cols].sort_values("final_regret"))
```
